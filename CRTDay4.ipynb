```python
names=["Ram","Raj","Rose","Jack","Pavi","Nikhil"]
marks=[[30,40,50],[60,77,55],[33,78,90],[87,97,57,],[55,66,77]]
for i in range(5):
  p=sum(marks[i])//3
  if p>80:
    g="F"
  elif p<40:
    g="F"
  elif p>40 and p<60:
    g="B"
  else:
    g="A"
  print("{}.Mr/Ms {} has scored {}% - {} Grade".format(i+1,names[i].title(),p ,g))
```

```python
import random
a=int(input("Enter num of teams:"))
teams=[]
print("Enter team names:")
for i in range(a):
  t=input("Enter team name: ")
  teams.append(t)
m=int(input("Enter num of meet bw teams: "))


matches=[]
for i in range(0,a-1):
  for j in range(i+1,a):
    for k in range(m):
      matches.append([(teams[i],teams[j])])
random.shuffle(matches)
pos = 1
for i in matches:
  print("MATCH {} : {} vs {}".format(pos, i[0][0], i[0][1]))
  pos+= 1
```
